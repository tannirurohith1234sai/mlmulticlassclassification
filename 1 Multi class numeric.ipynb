### OneVsRestClassifier

```python
import numpy as np

from sklearn.multiclass import OneVsRestClassifier

from sklearn.svm import SVC
```

```python

X = np.array([ [10, 10], [8, 10], [-5, 5.5], [-5.4, 5.5], [-20, -20], [-15, -20]  ])

y = np.array([0, 0, 1, 1, 2, 2])
```

```python
X
```

```python
y
```

```python
model = SVC()

ovr = OneVsRestClassifier(model)

clf = ovr.fit(X, y)
```

```python
clf
```

```python
clf.predict([[-19, -20], 
             [9, 9], 
             [-5, 5]])
```

### For each sample, all three classifiers give a score.

Example : Sample = [-19,-20]

    Classifier 0 score = -2
    
    Classifier 1 score = -1
    
    Classifier 2 score = +5

### Highest score wins.

### Prediction = Class 2

Similarly,

[9,9]
    
    Classifier0 = +4
    Classifier1 = -2
    Classifier2 = -5

    Prediction = 0

[-5,5]
    
    Classifier0 = -3
    Classifier1 = +4
    Classifier2 = -1

    Prediction = 1

### OneVsOneClassifier

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

from sklearn.multiclass import OneVsOneClassifier

from sklearn.svm import LinearSVC
```

### What happens internally?

Instead of One vs Rest

it creates

    Class 0 vs Class 1
    
    Class 0 vs Class 2
    
    Class 1 vs Class 2

Only two classes are used in each classifier.

| Classes | OvO Classifiers |
| ------- | --------------- |
| 3       | 3               |
| 4       | 6               |
| 5       | 10              |
| 6       | 15              |
| 10      | 45              |


```python
X, y = load_iris(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.33, shuffle=True, random_state=0)
```

```python
base_model = LinearSVC(dual="auto", random_state=0)

clf = OneVsOneClassifier(base_model)
clf.fit(X_train, y_train)
```

```python
clf.predict(X_test[:3])
```

| Feature                      | One-vs-Rest (OvR)                  | One-vs-One (OvO)             |
| ---------------------------- | ---------------------------------- | ---------------------------- |
| Classifier count             | k                                  | k(k−1)/2                     |
| Training data per classifier | All samples                        | Only two classes             |
| Prediction                   | Highest score                      | Majority voting              |
| Training speed               | Usually faster                     | Usually slower (more models) |
| Memory usage                 | Lower                              | Higher                       |
| Common with                  | Logistic Regression, Linear Models | SVM                          |
